## 5. Probar el agente en los cinco lagos

Entrena vuestro agente con 20 000 episodios y lo examina como en la evaluación: 1000 partidas sin explorar, eligiendo siempre la acción con mayor valor. Muestra la **recompensa media por partida** (de 0 a 2) y el tiempo.

- Con la plantilla vacía saldrá 0 o casi 0 en todos los lagos: el agente todavía no aprende nada (alguna vez llega por casualidad, al resbalar).
- Probad muchas semillas distintas: en la evaluación, la salida, la meta y las monedas estarán en posiciones que no conocéis.
- Cada lago tiene que tardar **menos de 5 minutos**.

```python
import time, importlib
import numpy as np
import agente
importlib.reload(agente)              # usa siempre la última versión de agente.py
from agente import entrenar
from lago_monedas import MAPAS, crear_lago

def examen(Q, entorno, partidas=1000):
    entorno.reset(seed=123)
    total = 0
    for _ in range(partidas):
        estado, _ = entorno.reset()
        fin = False
        while not fin:
            estado, r, terminado, truncado, _ = entorno.step(int(np.argmax(Q[estado])))
            fin = terminado or truncado
            total += r
    return total / partidas            # recompensa media por partida (0 a 2)

EPISODIOS = 20_000
SEMILLAS = [1, 2, 3]                   # cambiadlas y añadid más

for lago in range(5):
    for semilla in SEMILLAS:
        entorno = crear_lago(MAPAS[lago], semilla)
        inicio = time.time()
        Q = entrenar(entorno, EPISODIOS, semilla=0)
        print(f"Lago {lago + 1}, semilla {semilla}: recompensa media {examen(Q, entorno):.2f}"
              f" · {time.time() - inicio:.0f} s")
```
